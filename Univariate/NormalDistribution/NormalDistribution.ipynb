# Prob Dist Function

```python
def pdfGraph(ds,x1, x2):
    import seaborn as sns
    from scipy.stats import norm
    import matplotlib.pyplot as plt

    ax = sns.distplot(ds,kde=True,kde_kws={'color':'Blue'},color='Grey')
    plt.axvline(x1, color='Green')
    plt.axvline(x2, color='Red')
    plt.show()


    data=ds
    mean = data.mean()
    std = data.std()
    print('Mean=%.3f,Standard Deviation=%.3f' % (mean,std))
# Define Distribution
    dist = norm(mean,std)

    #Range outcome 
    values = [value for value in range(x1,x2)]
    prob = [dist.pdf(value) for value in values]
    probability = sum(prob)
    print("The area between the range({},{}):{}".format(x1,x2,sum(prob)))
    return probability  
   
```

```python
pdfGraph(ds["ssc_p"],35, 45)
```
